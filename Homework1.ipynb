Bai 2:Một khu đất hình chữ nhật có chiều rộng a, chiều dài b. Người ra xây một khu vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào a, b. Hãy tính phần diện tích còn để trồng cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2 chữ số phần thập phân.


```python
a=float(input("chieu dai cua khu dat:"))
b=float(input("chieu rong cua khu dat:"))
r=b/2
s1=a*b
s2=3.14*r**2
s3=s1-s2
s4=round(s3,2)
print("dien tich cua khu dat la:",s4)


```

Bài 4: Viết chương trình nhập vào một “kí tự”, Kiểm tra xem c có phải là kí tự chữ cái không (hoa hoặc thường). Nếu đúng thì in ra “{c} là kí tu alphabet”, nếu sai in ra “{c} 
không phải là kí tự alphabet. 

```python
c=input("nhap mot ki tu:")
b=ord(c)
if 65<=b<=122:
    chr(b)==True
    print("ki tu la alphabet")
else:
    chr(b)==False
    print("ki tu khong phai la alphabet")
```
