```python
a=float(input("chieu dai cua khu dat:"))
b=float(input("chieu rong cua khu dat:"))
r=b/2
s1=a*b
s2=3.14*r**2
s3=s1-s2
s4=round(s3,2)
print("dien tich cua khu dat la:",s4)


```

Bài 4: Viết chương trình nhập vào một “kí tự”, Kiểm tra xem c có phải là kí tự chữ cái không (hoa hoặc thường). Nếu đúng thì in ra “{c} là kí tu alphabet”, nếu sai in ra “{c}

```python
c=input("nhap mot ki tu:")
b=ord(c)
if 65<=b<=122:
    chr(b)==True
    print("ki tu la alphabet")
else:
    chr(b)==False
    print("ki tu khong phai la alphabet")
```

Bài 6. Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, 
nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó 
không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”.

```python
a=float(input("nhap vao so thuc:"))
b=float(input("nhap vao so thuc:"))
c=float(input("nhap vao so thuc:"))
if a>0 and b>0 and c>0:
    if a+b>c and a+c>b and b+c>a:
        p=(a+b+c)/2
        s=(p*(p-a)*(p-b)*(p-c))**0.5
        print("dien tich cua tam giac la:",round(s,1))
    else:
        print("khong phai la tam giac")
```
