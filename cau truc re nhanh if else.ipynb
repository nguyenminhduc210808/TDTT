Nhập một tháng (tây), cho biết tháng đó thuộc mùa nào (Xuân, Hạ, Thu, Ðông), biết 
rằng tháng 2, 3, 4 là mùa Xuân, tháng 5, 6, 7: mùa Hạ, tháng 8, 9, 10: mùa Thu, và 
tháng 11, 12, 1: mùa Ðông.

```python
n=int(input())
match n:
    case x if x<0:
        print(f"{n} la so am")
    case z if z>0:
        print(f"{n} la so duong")
    case _:
        print("Error")
```
