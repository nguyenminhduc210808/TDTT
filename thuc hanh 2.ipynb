Viết chương trình nhập vào tên 3 người từ bàn phím.  
In ra màn hình tên theo thứ tự đảo ngược. 
VD: input ("Bob Alice Helen") thì output: ("Hi Helen, Alice and Bob.")


```python
names=input()
print(type(names))
ten=names.split()
print(type(ten))
print(f"Hello {ten[2]}, {ten[1]}, {ten[0]}")
```

```python
names=input()
print(type(names))
ten=names.split(sep="=")
print(type(ten))
print(f"Hello {ten[2]}, {ten[1]}, {ten[0]}")
```
